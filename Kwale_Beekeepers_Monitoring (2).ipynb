```python
hives = [
    {"hive_id": "A01",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 10,
        "days_since_inspection": 5
     },
    { "hive_id": "A02",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 7,
        "days_since_inspection": 9
      },
   { "hive_id": "A07",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 9.5,
        "days_since_inspection": 10
     },
    { "hive_id": "A04",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 8,
        "days_since_inspection": 6
      },
    { "hive_id": "A05",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 11,
        "days_since_inspection": 5
      },

   {"hive_id": "A03",
        "colony_active": True,
        "pests_detected": False,
        "honey_kg": 5,
        "days_since_inspection": 8
    }
]

print("Number of hive records:", len(hives))
```

```python
import pandas as pd

df = pd.DataFrame(hives)

df
```

```python

# Count missing hive IDs
print("Missing hive IDs:", df["hive_id"].isna().sum())

# Check for duplicate hive IDs
print("Duplicate hive IDs:", df["hive_id"].duplicated().sum())

# Check for negative honey quantities
print("Negative honey values:", (df["honey_kg"] < 0).sum())
```

```python

def check_hive(row):
    issues = []

    if row["pests_detected"]:
        issues.append("Inspect reported pest issue")

    if not row["colony_active"]:
        issues.append("Verify colony activity")

    if row["days_since_inspection"] > 14:
        issues.append("Schedule overdue inspection")

    if issues:
        return "; ".join(issues)

    return "No immediate flag"


df["recommended_action"] = df.apply(check_hive, axis=1)

df[["hive_id", "recommended_action"]]
```

```python

print("Total hives:", len(df))

print(
    "Active colonies:",
    int(df["colony_active"].sum())
)

print(
    "Hives requiring follow-up:",
    int((df["recommended_action"] != "No immediate flag").sum())
)

print(
    "Total honey recorded (kg):",
    df["honey_kg"].sum()
)
```

```python
test_hive = {
    "hive_id": "TEST1",
    "colony_active": True,
    "pests_detected": True,
    "honey_kg": 2.0,
    "days_since_inspection": 5
}

# Test the follow-up function
print(check_hive(test_hive))
```
